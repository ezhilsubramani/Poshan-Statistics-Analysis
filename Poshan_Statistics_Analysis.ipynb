## Stage 2 – Data Cleaning and Pre-processing



**Handle missing values (impute or drop)**

```python

missing_data=pd.DataFrame({
    'missing_values':df.isnull().sum(),
    'missing_percentages':((df.isnull().sum()/len(df))*100).round(2)
    })
missing_data=missing_data.sort_values(by='missing_percentages',ascending=False)
missing_data
```

```python
low_percentage=missing_data[missing_data['missing_percentages']<5.0].index
low_percentage
for col in low_percentage:
 if df[col].dtype !='object':
  df[col]=df[col].fillna(df[col].median())
```

```python
high_percentage=missing_data[missing_data['missing_percentages']>5.0].index
high_percentage
for col in high_percentage:
 if df[col].dtype !='object':
  df[col]=df[col].fillna(0)
```

**Box Plot**

```python

# Select numerical columns
numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns

# Plot boxplots
plt.figure(figsize=(20, 8))
sns.boxplot(data=df[numerical_cols])

plt.xticks(rotation=90)
plt.title("Box Plot of Numerical Columns (Before Outlier Treatment)")
plt.show()
```

**Observation**

A large number of outliers were observed across multiple numerical variables.

Beneficiary and service-related columns showed the highest variability.

The distributions were positively skewed with several extreme values.

Therefore, the IQR (Interquartile Range) method was applied to identify and cap extreme values while preserving all observations.

**Interquartile Range**

```python
# Step 1: Select numerical columns
numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns

# Step 2: Exclude identifier columns
exclude_cols = ['id', 'state_code', 'district_code']

numerical_cols = [col for col in numerical_cols if col not in exclude_cols]

# Step 3: Create a copy of the dataset
df_cap = df.copy()

# Step 4: Apply IQR Outlier Capping
for col in numerical_cols:

    # Calculate Quartiles
    Q1 = df_cap[col].quantile(0.25)
    Q3 = df_cap[col].quantile(0.75)

    # Calculate IQR
    IQR = Q3 - Q1

    # Calculate Lower and Upper Limits
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    # Cap Outliers
    df_cap[col] = df_cap[col].clip(lower=lower_limit, upper=upper_limit)

# Step 5: Replace original dataset
df = df_cap.copy()

print("IQR Outlier Capping Completed Successfully!")
```

# **5.Which regions have poor infrastructure facilities?**

```python
# Group infrastructure data by state
infra = df.groupby('state_name')[
    [
        'awc_infra_own_building',
        'awc_infra_func_toilets',
        'awc_infra_drink_water_src'
    ]
].sum()

# Calculate total infrastructure
infra['Total Infrastructure'] = (
    infra['awc_infra_own_building'] +
    infra['awc_infra_func_toilets'] +
    infra['awc_infra_drink_water_src']
)

# Select 10 states with the LOWEST infrastructure
top10 = infra.sort_values(
    by='Total Infrastructure',
    ascending=True
).head(10)

# Keep only infrastructure columns for the heatmap
heatmap_data = top10[
    [
        'awc_infra_own_building',
        'awc_infra_func_toilets',
        'awc_infra_drink_water_src'
    ]
]

# Rename columns for better readability
heatmap_data.columns = [
    'Own Building',
    'Functional Toilets',
    'Drinking Water'
]

# Plot heatmap
plt.figure(figsize=(10, 6))

sns.heatmap(
    heatmap_data,
    annot=True,
    fmt=".0f",
    cmap="YlGnBu",
    linewidths=0.5
)

plt.title("10 States with the Lowest Anganwadi Infrastructure Availability")
plt.xlabel("Infrastructure Facilities")
plt.ylabel("States")

plt.tight_layout()
plt.show()
```

**Analysis Type:**
Diagnostic Analysis

**Chart:**

**Columns:**

awc_infra_own_building,
awc_infra_func_toilets,

**Insights:**
